```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("CodeAlpha_ECommerce_Sales_Dataset.xlsx")

df.head()
```

```python
monthly_revenue = df.groupby("Month")["Revenue"].sum()

month_order = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

monthly_revenue = monthly_revenue.reindex(month_order)

plt.figure(figsize=(10, 5))
plt.plot(monthly_revenue.index, monthly_revenue.values, marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_revenue = df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(category_revenue.index, category_revenue.values)

plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
city_revenue = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(city_revenue.index, city_revenue.values)

plt.title("Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
plt.barh(top_products.index[::-1], top_products.values[::-1])

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
quantity_revenue = df.groupby("Quantity")["Revenue"].sum()

plt.figure(figsize=(8, 5))
plt.bar(quantity_revenue.index.astype(str), quantity_revenue.values)

plt.title("Revenue by Quantity")
plt.xlabel("Quantity")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
quarter_revenue = df.groupby("Quarter")["Revenue"].sum()

plt.figure(figsize=(8, 5))
plt.bar(quarter_revenue.index, quarter_revenue.values)

plt.title("Revenue by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Monthly Revenue
monthly_revenue = df.groupby("Month")["Revenue"].sum()
month_order = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]
monthly_revenue = monthly_revenue.reindex(month_order)

axes[0, 0].plot(monthly_revenue.index, monthly_revenue.values, marker="o")
axes[0, 0].set_title("Monthly Revenue Trend")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Revenue")
axes[0, 0].tick_params(axis="x", rotation=45)

# 2. Category Revenue
category_revenue = df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)

axes[0, 1].bar(category_revenue.index, category_revenue.values)
axes[0, 1].set_title("Revenue by Category")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Revenue")

# 3. City Revenue
city_revenue = df.groupby("City")["Revenue"].sum().sort_values(ascending=False)

axes[1, 0].bar(city_revenue.index, city_revenue.values)
axes[1, 0].set_title("Revenue by City")
axes[1, 0].set_xlabel("City")
axes[1, 0].set_ylabel("Revenue")
axes[1, 0].tick_params(axis="x", rotation=45)

# 4. Quarter Revenue
quarter_revenue = df.groupby("Quarter")["Revenue"].sum()

axes[1, 1].bar(quarter_revenue.index, quarter_revenue.values)
axes[1, 1].set_title("Revenue by Quarter")
axes[1, 1].set_xlabel("Quarter")
axes[1, 1].set_ylabel("Revenue")

plt.suptitle("E-Commerce Sales Visualization Dashboard", fontsize=16)
plt.tight_layout()
plt.show()
```
